```python
import pandas as pd
from pathlib import Path
import json
```

```python
def get_latest_file(folder, glob_str):
    files = list(folder.glob(glob_str))
    return sorted(files)[-1]
```

### Ghostfolio Activities

```python
ghostfolio_folder = Path("/Users/veselda7/Library/CloudStorage/GoogleDrive-veselda7@gmail.com/My Drive/S Kačenkou/Investovani/ghostfolio")

latest_ghostfolio_file = get_latest_file(ghostfolio_folder, "*.json")
print(latest_ghostfolio_file.name)
```

```python
with latest_ghostfolio_file.open("r", encoding="utf-8") as f:
    data = json.load(f)
```
